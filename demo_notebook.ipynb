# NOAA Significant US Storms (1950-2025)

Explore **14,770 significant weather events** curated from 75 years of NOAA data.

**Dataset Highlights:**
- F3/EF3+ tornadoes, Category 3+ hurricanes
- Fatal events and $1M+ damage storms
- 98.8% reduction from 1.23M original records

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-whitegrid')
print('Libraries loaded')
```

## 1. Load Dataset

```python
with open('noaa_significant_storms.json') as f:
    data = json.load(f)

df = pd.DataFrame(data)
print(f'Total storms: {len(df):,}')
df.head()
```

## 2. Event Type Distribution

```python
event_counts = df['event_type'].value_counts()
print('Event Types:')
print('=' * 40)
for event, count in event_counts.head(10).items():
    pct = count / len(df) * 100
    print(f'{event:25s} {count:6,} ({pct:5.1f}%)')
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
event_counts.head(10).plot(kind='barh', ax=ax, color='steelblue')
ax.set_xlabel('Number of Events')
ax.set_title('Top 10 Storm Event Types', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Geographic Distribution

```python
state_counts = df['state'].value_counts().head(15)
print('Top 15 States:')
print('=' * 40)
for state, count in state_counts.items():
    print(f'{state:20s} {count:6,}')
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
ax.scatter(df['longitude'], df['latitude'], alpha=0.3, s=2, c='red')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title('Storm Locations Across the US', fontweight='bold')
ax.set_xlim(-130, -65)
ax.set_ylim(24, 50)
plt.tight_layout()
plt.show()
```

## 4. Temporal Patterns

```python
df['year'] = pd.to_datetime(df['date']).dt.year
yearly = df.groupby('year').size()

fig, ax = plt.subplots(figsize=(14, 5))
yearly.plot(ax=ax, color='darkblue')
ax.set_xlabel('Year')
ax.set_ylabel('Number of Significant Storms')
ax.set_title('Significant Storm Events by Year (1950-2025)', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. Fatalities and Damage

```python
fatal = df[df['fatalities'] > 0]
print(f'Events with fatalities: {len(fatal):,}')
print(f'Total fatalities: {df["fatalities"].sum():,}')
print(f'Total injuries: {df["injuries"].sum():,}')
print(f'Total property damage: ${df["damage_property"].sum()/1e9:.1f} billion')
```

```python
deadliest = df.nlargest(10, 'fatalities')[['date', 'state', 'event_type', 'fatalities', 'injuries']]
print('\nDeadliest Events:')
print('=' * 70)
print(deadliest.to_string(index=False))
```

## 6. Tornado Analysis

```python
tornadoes = df[df['event_type'] == 'Tornado']
print(f'Total tornadoes: {len(tornadoes):,}')

if 'magnitude' in tornadoes.columns:
    mag_counts = tornadoes['magnitude'].value_counts().sort_index()
    print('\nTornado Ratings:')
    for mag, count in mag_counts.items():
        print(f'{mag}: {count:,}')
```

## Conclusion

This notebook demonstrated:

- Event type distribution (tornadoes dominate)
- Geographic patterns (Texas leads)
- Temporal trends over 75 years
